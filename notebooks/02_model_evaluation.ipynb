# 02 - Model Evaluation

Load trained artifacts and inspect metrics on the time-based test split.
Train first with `python scripts/train_models.py --config configs/base.yaml`.

```python
import sys, os
sys.path.insert(0, os.path.abspath(os.path.join('..', 'src')))

from football_predictor.data import loaders
from football_predictor.evaluation.metrics import evaluate_1x2
from football_predictor.prediction.predictor import load_artifacts, predict_proba_df
from football_predictor.utils.config import load_config
from football_predictor.utils.dates import time_split

cfg = load_config('../configs/base.yaml')
df = loaders.load_processed(cfg.data.processed_features_parquet)
artifacts = load_artifacts(cfg.model.save_dir, cfg)
v = cfg.validation
test = time_split(df, v.train_end_date, v.validation_start_date, v.validation_end_date, v.test_start_date)['test']
test = test[test['result_1x2'].notna()]
proba = predict_proba_df(artifacts, test)
evaluate_1x2(test['result_1x2'].to_numpy(), proba)
```
